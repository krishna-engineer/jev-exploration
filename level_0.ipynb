```python
import os
import time

from typesafe_sdk import Choice, Noul, Score, TypeSafeClient
```

```python
typesafe_api_key =  os.getenv('typesafe_api_key')
```

```python
client = TypeSafeClient(api_key=typesafe_api_key,
                        model="jev-latest")
```

```python
state = """
This person is an AI and machine learning professional who has worked at the same company for about 7 years. Their role has gradually shifted from hands-on engineering toward leading a team and translating product requirements into technical work.

The engineers who joined the team are highly capable. This person now spends much of their time helping them understand priorities, resolving ambiguity, and connecting their work to what the product needs. They value that contribution, but miss building things themselves and having enough uninterrupted time to study a technical problem in depth.

They recognize that “I don’t have time” may partly reflect their own habits or bias. Still, the shift in how they spend their working days is real. They are considering roles where they can continue guiding technical decisions while doing more hands-on work and developing deeper expertise.
"""
```

```python
questions = {
    "source_of_reduced_hands_on_work": Choice(
        instructions=(
            "Based only on the state, what best explains the person's "
            "reduced hands-on engineering time?"
        ),
        criteria={
            "role_demands": "Leadership and coordination take up the time.",
            "personal_habits": "The person's own use of time is the main cause.",
            "both": "Both role demands and personal habits contribute.",
            "insufficient_evidence": "The state cannot distinguish the causes.",
        },
    ),
    "role_mismatch": Score(
        instructions=(
            "How much does the person's current work differ from the work "
            "they say they want to do?"
        ),
        criteria=[
            "Little or no mismatch.",
            "Some mismatch, but desired work remains possible.",
            "Clear mismatch that affects regular work.",
            "Substantial mismatch: desired work is largely absent.",
        ],
    ),
    "next_step_to_investigate": Choice(
        instructions=(
            "Which single next step would most help the person understand "
            "or address the situation described?"
        ),
        criteria={
            "reshape_current_role": "Discuss a change in responsibilities.",
            "test_protected_time": "Reserve technical time and observe whether it works.",
            "explore_new_roles": "Investigate roles with explicit hands-on work.",
            "gather_more_evidence": "First record how time is spent and what blocks deep work.",
        },
    ),
    "is_resigning_good_decision": Noul(
        instructions="Think through industry standards and let me know if resigning from current company would be right decision"
    )
}
```

```python

for i in range(20):
    start_time = time.perf_counter()
    resp = client.system_one(
        state=state,
        questions=questions
    )

    print(f"Code took {time.perf_counter() - start_time:.4f} seconds to run.")
```
